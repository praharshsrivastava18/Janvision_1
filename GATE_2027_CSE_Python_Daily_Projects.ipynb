## Day 01 — Discrete mathematics: relations and graph traversal

**GATE area:** Engineering Mathematics + Algorithms  
**Syllabus link:** Sets, relations, graph connectivity, and graph traversal.

Represent a graph as an adjacency list. Breadth-first search (BFS) uses a queue so vertices are visited in nondecreasing distance from the start.

**Picture:** A BFS wave spreads through nearby stations first, then reaches the stations one transfer farther away.

**Python focus:** deque.popleft() gives an efficient queue; a set makes membership checks efficient.  
**Java focus:** Queue, ArrayDeque, List, Map, and HashSet play the same roles as Python deque, list, and set.

**Complexity:** With adjacency lists, BFS takes O(V + E) time and O(V) extra space.

**Trap:** A graph may be directed or undirected; encode edges consistently. A visited set prevents cycles from endlessly re-adding vertices.

**Try next:** Add a disconnected vertex and write a component-count routine. Explain why a vertex is marked seen when it is enqueued, not when it is removed.

Matching program: `java/Day01_*.java`.


````python
from collections import deque

graph = {
    0: [1, 2],
    1: [0, 3],
    2: [0, 3],
    3: [1, 2, 4],
    4: [3],
}

def bfs(graph, start):
    seen = {start}
    order = []
    queue = deque([start])
    while queue:
        node = queue.popleft()
        order.append(node)
        for neighbour in graph[node]:
            if neighbour not in seen:
                seen.add(neighbour)
                queue.append(neighbour)
    return order

order = bfs(graph, 0)
assert order == [0, 1, 2, 3, 4]
print("BFS order:", order)
````

**Expected check:** BFS order: [0, 1, 2, 3, 4]

## Day 02 — Linear algebra and probability lab

**GATE area:** Engineering Mathematics  
**Syllabus link:** Matrices, systems of equations, conditional probability, and Bayes theorem.

Matrix multiplication combines linear transformations. Bayes theorem reverses a conditional probability when the base rate is known.

**Picture:** A matrix is a small machine that transforms an input vector; Bayes updates an initial belief after new evidence arrives.

**Python focus:** Nested list comprehensions and sum() express the row-by-column dot product. Use floats for probabilities.  
**Java focus:** Two-dimensional arrays use row/column indices; double preserves fractional probability values.

**Complexity:** Multiplying two n x n matrices with the direct method takes O(n^3) time. The displayed probability calculation is O(1).

**Trap:** The order of matrix multiplication matters. In Bayes problems, include both true positives and false positives in the denominator.

**Try next:** Change the prior prevalence to 0.10 and compare the posterior. Verify that A (2x3) times B (3x2) is defined, but the reverse product is not necessarily defined.

Matching program: `java/Day02_*.java`.


````python
def multiply(a, b):
    return [[sum(a[i][k] * b[k][j] for k in range(len(b)))
             for j in range(len(b[0]))]
            for i in range(len(a))]

A = [[1, 2], [3, 4]]
B = [[2, 0], [1, 2]]
product = multiply(A, B)
assert product == [[4, 4], [10, 8]]

prevalence = 0.01
sensitivity = 0.95
false_positive = 0.05
posterior = (sensitivity * prevalence) / (
    sensitivity * prevalence + false_positive * (1 - prevalence))
assert round(posterior, 4) == 0.161
print("A x B:", product)
print("P(condition | positive):", round(posterior, 4))
````

**Expected check:** A x B: [[4, 4], [10, 8]]; P(condition | positive): 0.1610

## Day 03 — Digital logic: truth tables and adders

**GATE area:** Digital Logic  
**Syllabus link:** Boolean algebra, combinational circuits, and fixed-width number representation.

Build a truth table from Boolean operators, then implement a full-adder output using XOR for sum and majority logic for carry.

**Picture:** A logic circuit is a tiny decision machine: input bits arrive, and gates combine them into output bits.

**Python focus:** range/loops enumerate every input combination; bitwise ~ must be limited with a width mask.  
**Java focus:** Integer bitwise operators use 32-bit ints; mask with (1 << bits) - 1 to demonstrate a smaller width.

**Complexity:** Each of the 8 full-adder rows is constant work; generating the table is O(1) for a fixed 3-bit input.

**Trap:** A bitwise complement depends on a fixed width; mask the result to the desired number of bits. Unsigned and signed interpretations differ.

**Try next:** Write the truth table for XOR and XNOR. What is the signed integer range for n-bit two’s complement?

Matching program: `java/Day03_*.java`.


````python
def full_adder(a, b, carry_in):
    total = a + b + carry_in
    return total % 2, total // 2

rows = []
for a in (0, 1):
    for b in (0, 1):
        for carry_in in (0, 1):
            rows.append((a, b, carry_in, *full_adder(a, b, carry_in)))
assert rows[-1] == (1, 1, 1, 1, 1)
assert all(a + b + c == s + 2 * carry
           for a, b, c, s, carry in rows)

def twos_complement(value, bits):
    return ((~value) + 1) & ((1 << bits) - 1)

assert twos_complement(5, 8) == 251  # 11111011 = -5 in 8 bits
print("(a,b,cin,sum,cout):", rows)
print("-5 in 8-bit two's complement:", format(twos_complement(5, 8), "08b"))
````

**Expected check:** Full-adder rows enumerate all 8 inputs; 8-bit two’s complement of 5 is 11111011.

## Day 05 — Trees, binary search trees, and heaps

**GATE area:** Programming and Data Structures  
**Syllabus link:** Trees, binary search trees, and binary heaps.

A BST orders keys so left < node < right; inorder traversal produces sorted keys. A min-heap exposes the smallest element quickly.

**Picture:** A BST is a sequence of yes/no turns that narrows a search; a heap keeps the next smallest priority at the top.

**Python focus:** Recursive insertion mirrors the tree definition; heapq provides a min-heap using a list.  
**Java focus:** PriorityQueue provides heap ordering. The BST node class stores child references; traversal writes into a List.

**Complexity:** BST insertion/search is O(h), where h is height. Heap insertion/removal is O(log n); peek-min is O(1).

**Trap:** BST operations depend on tree height: O(log n) only when balanced; the worst case is O(n). A heap is not a fully sorted array.

**Try next:** Insert an already sorted sequence and draw the BST. Explain why a plain BST can become a chain, and how a balanced BST changes the height.

Matching program: `java/Day05_*.java`.


````python
import heapq

class Node:
    def __init__(self, key):
        self.key, self.left, self.right = key, None, None

def insert(root, key):
    if root is None:
        return Node(key)
    if key < root.key:
        root.left = insert(root.left, key)
    elif key > root.key:
        root.right = insert(root.right, key)
    return root

def inorder(root):
    if root is None:
        return []
    return inorder(root.left) + [root.key] + inorder(root.right)

root = None
for key in [8, 3, 10, 1, 6]:
    root = insert(root, key)
assert inorder(root) == [1, 3, 6, 8, 10]
heap = [8, 3, 10, 1, 6]
heapq.heapify(heap)
assert heapq.heappop(heap) == 1
print("BST inorder:", inorder(root), "| heap minimum: 1")
````

**Expected check:** BST inorder: [1, 3, 6, 8, 10] | heap minimum: 1

## Day 06 — Sorting, searching, and complexity

**GATE area:** Algorithms  
**Syllabus link:** Searching, sorting, worst-case time complexity, and divide-and-conquer.

Merge sort divides a list, sorts each half, and merges them. Binary search halves a sorted search interval each step.

**Picture:** Merge sort sorts two smaller piles before merging; binary search finds a word by repeatedly opening the middle of a dictionary.

**Python focus:** Slices create new lists in this teaching implementation; production implementations can reduce copying.  
**Java focus:** Arrays.copyOfRange makes the divide steps clear; primitive int[] avoids boxing for this example.

**Complexity:** Merge sort: O(n log n) time and O(n) extra space. Binary search: O(log n) time and O(1) extra space.

**Trap:** Binary search requires sorted data. Use low + (high - low) / 2 to avoid integer overflow in fixed-width languages.

**Try next:** Count comparisons for binary search on 16 sorted keys. What changes if the input is not sorted? Compare merge sort and insertion sort for nearly sorted input.

Matching program: `java/Day06_*.java`.


````python
def merge_sort(values):
    if len(values) <= 1:
        return values[:]
    mid = len(values) // 2
    left = merge_sort(values[:mid])
    right = merge_sort(values[mid:])
    merged = []
    while left and right:
        merged.append(left.pop(0) if left[0] <= right[0] else right.pop(0))
    return merged + left + right

def binary_search(values, target):
    low, high = 0, len(values) - 1
    while low <= high:
        mid = (low + high) // 2
        if values[mid] == target:
            return mid
        if values[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

ordered = merge_sort([8, 3, 6, 1, 5])
assert ordered == [1, 3, 5, 6, 8]
assert ordered[binary_search(ordered, 6)] == 6
print("sorted:", ordered, "| index of 6:", binary_search(ordered, 6))
````

**Expected check:** sorted: [1, 3, 5, 6, 8] | index of 6: 3

## Day 07 — Greedy, dynamic programming, and graph algorithms

**GATE area:** Algorithms  
**Syllabus link:** Greedy algorithms, dynamic programming, shortest paths, and minimum spanning trees.

Different objectives call for different strategies: earliest-finish selection, DP for 0/1 knapsack, Dijkstra for nonnegative shortest paths, and Kruskal for MST.

**Picture:** Greedy picks the best next local move; DP keeps answers to overlapping subproblems; MST connects every place as cheaply as possible.

**Python focus:** heapq supports Dijkstra’s priority queue; zip pairs each weight with its value.  
**Java focus:** PriorityQueue orders distance/node pairs; arrays keep the DP table and distance labels compact.

**Complexity:** Activity selection O(n log n); 0/1 knapsack O(nW); Dijkstra with a binary heap O((V+E) log V).

**Trap:** Dijkstra requires nonnegative edge weights. In 0/1 knapsack, capacities must be traversed downward so an item cannot be reused in the same iteration.

**Try next:** For the same knapsack, compare a greedy value/weight choice with the exact DP result. Add Kruskal’s algorithm using a disjoint-set union structure.

Matching program: `java/Day07_*.java`.


````python
import heapq

def activity_count(activities):
    chosen, finish = [], float("-inf")
    for start, end in sorted(activities, key=lambda item: item[1]):
        if start >= finish:
            chosen.append((start, end))
            finish = end
    return chosen

def knapsack(weights, values, capacity):
    best = [0] * (capacity + 1)
    for weight, value in zip(weights, values):
        for cap in range(capacity, weight - 1, -1):
            best[cap] = max(best[cap], best[cap - weight] + value)
    return best[capacity]

def dijkstra(graph, source):
    dist = {v: float("inf") for v in graph}
    dist[source] = 0
    heap = [(0, source)]
    while heap:
        cost, node = heapq.heappop(heap)
        if cost != dist[node]:
            continue
        for nxt, weight in graph[node]:
            new_cost = cost + weight
            if new_cost < dist[nxt]:
                dist[nxt] = new_cost
                heapq.heappush(heap, (new_cost, nxt))
    return dist

activities = activity_count([(1, 3), (2, 5), (4, 7), (6, 9)])
assert activities == [(1, 3), (4, 7)]
assert knapsack([2, 3, 4], [4, 5, 7], 5) == 9
graph = {0: [(1, 4), (2, 1)], 1: [(0, 4), (2, 2), (3, 1)],
         2: [(0, 1), (1, 2), (3, 5)], 3: [(1, 1), (2, 5)]}
assert dijkstra(graph, 0)[3] == 4
print("activities:", activities, "| knapsack:", 9, "| dist to 3:", 4)
````

**Expected check:** activities: [(1, 3), (4, 7)] | knapsack: 9 | dist to 3: 4

**Exam note:** Python/Java are practice languages here; make sure you can explain the same idea, trace C/pseudocode, and solve a paper question.


## Day 08 — Operating systems: CPU scheduling and deadlock safety

**GATE area:** Operating Systems  
**Syllabus link:** Processes, CPU scheduling, synchronization, and deadlock.

FCFS computes waiting time from the prior completion time. A Banker-style safety check searches for an order in which each process can finish with currently available resources.

**Picture:** A CPU is a single service desk; a deadlock-safe sequence is a queue order where each customer can finish and return borrowed resources.

**Python focus:** Lists store waiting times and resource vectors. next() finds the first process whose needs fit.  
**Java focus:** Primitive arrays represent one resource type clearly; boolean[] marks finished processes.

**Complexity:** FCFS calculation is O(n). The simple safe-sequence search is O(n^2 r) for n processes and r resource types.

**Trap:** A safe state is not the same as a deadlocked state; it means a completion order exists. FCFS can cause a convoy effect.

**Try next:** Calculate turnaround time for each FCFS process. Change available resources to zero and identify whether a safe sequence exists.

Matching program: `java/Day08_*.java`.


````python
def fcfs(bursts):
    clock, waits = 0, []
    for burst in bursts:
        waits.append(clock)
        clock += burst
    return waits, clock

def safe_sequence(available, allocation, need):
    work = available[:]
    done = [False] * len(need)
    sequence = []
    while len(sequence) < len(need):
        candidate = next((i for i in range(len(need))
                          if not done[i] and all(n <= w for n, w in zip(need[i], work))), None)
        if candidate is None:
            return None
        work = [w + a for w, a in zip(work, allocation[candidate])]
        done[candidate] = True
        sequence.append(candidate)
    return sequence

waiting, finish = fcfs([5, 3, 2])
assert waiting == [0, 5, 8] and finish == 10
sequence = safe_sequence([1], [[1], [1]], [[2], [0]])
assert sequence == [1, 0]
print("FCFS waiting:", waiting, "| finish:", finish, "| safe sequence:", sequence)
````

**Expected check:** FCFS waiting: [0, 5, 8] | finish: 10 | safe sequence: [1, 0]

**Exam note:** Python/Java are practice languages here; make sure you can explain the same idea, trace C/pseudocode, and solve a paper question.

## Day 10 — Computer networks: IPv4 CIDR and parity

## Day 11 — Theory of computation: DFA and grammar recognition

**GATE area:** Theory of Computation  
**Syllabus link:** Regular expressions, finite automata, context-free grammars, pumping lemma, Turing machines, and undecidability.

A deterministic finite automaton tracks a finite state. A simple parser can recognize balanced brackets using a stack, which models more memory than a DFA.

**Picture:** A DFA is a small machine with a fixed number of moods; a stack machine remembers nested openings until they are closed.

**Python focus:** The DFA keeps only a two-character state; the bracket routine explicitly checks negative depth and final depth.  
**Java focus:** String state update shows finite memory; the integer depth is a simple stack-depth model for one bracket type.

**Complexity:** Both scans take O(n) time. Parenthesis recognition uses O(1) counter space; a general CFG parser may need more.

**Trap:** A regular expression describes a regular language; balanced arbitrary nesting is context-free, not regular. Test empty strings and malformed input.

**Try next:** Draw the DFA states for suffix 01 and transitions on 0/1. Give a grammar for balanced parentheses and explain why finite-state memory alone cannot count arbitrary nesting depth.

Matching program: `java/Day11_*.java`.


````python
def ends_with_01(bits):
    # DFA state is the last two bits seen; None means fewer than two.
    state = ""
    for bit in bits:
        if bit not in "01":
            return False
        state = (state + bit)[-2:]
    return state == "01"

def balanced_parentheses(text):
    depth = 0
    for char in text:
        if char == "(": depth += 1
        elif char == ")":
            depth -= 1
            if depth < 0: return False
    return depth == 0

assert ends_with_01("1101") and not ends_with_01("1110")
assert balanced_parentheses("(()())") and not balanced_parentheses("())(")
print("DFA 1101:", ends_with_01("1101"), "| balanced (()()):", balanced_parentheses("(()())"))
````

**Expected check:** DFA accepts 1101 and rejects 1110; balanced-parentheses parser accepts (()()).

## Day 12 — Compiler design: tokenizer and recursive-descent parser

**GATE area:** Compiler Design  
**Syllabus link:** Lexical analysis, parsing, syntax-directed translation, runtime environments, intermediate code, and local optimization.

A tokenizer turns characters into tokens. A recursive-descent parser follows grammar rules; emitting postfix while parsing is a small translation step.

**Picture:** A compiler first recognizes words, then checks sentence structure, then translates the sentence into a simpler instruction sequence.

**Python focus:** A regular expression splits simple tokens; the parser index makes its current position explicit.  
**Java focus:** A character scan handles multi-digit integers; List<String> stores tokens and the parser index tracks consumption.

**Complexity:** Tokenization and this parser are O(n) time for an expression of n characters/tokens.

**Trap:** The grammar encodes precedence: expr -> term (+/- term)*; term -> factor (*// factor)*. A parser must also reject leftover tokens.

**Try next:** Add exponentiation with the correct precedence and associativity. Reject a missing closing parenthesis and describe how a symbol table supports runtime environments.

Matching program: `java/Day12_*.java`.


````python
import re

def tokenize(expression):
    return re.findall(r"\d+|[()+*/-]", expression.replace(" ", ""))

class Parser:
    def __init__(self, tokens): self.tokens, self.i = tokens, 0
    def take(self):
        token = self.tokens[self.i]; self.i += 1; return token
    def factor(self):
        token = self.take()
        if token == "(":
            result = self.expr()
            assert self.take() == ")"
            return result
        assert token.isdigit()
        return [token]
    def term(self):
        out = self.factor()
        while self.i < len(self.tokens) and self.tokens[self.i] in "*/":
            op = self.take(); out += self.factor() + [op]
        return out
    def expr(self):
        out = self.term()
        while self.i < len(self.tokens) and self.tokens[self.i] in "+-":
            op = self.take(); out += self.term() + [op]
        return out

tokens = tokenize("2 + 3 * (4 - 1)")
postfix = Parser(tokens).expr()
assert postfix == ["2", "3", "4", "1", "-", "*", "+"]
print("tokens:", tokens, "| postfix:", postfix)
````

**Expected check:** tokens: [2, +, 3, *, (, 4, -, 1, )] | postfix: [2, 3, 4, 1, -, *, +]
